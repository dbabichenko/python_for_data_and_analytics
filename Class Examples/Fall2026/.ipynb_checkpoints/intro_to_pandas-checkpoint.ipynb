# Concat Examples

```python
df1 = df[['id', 'age', 'gender']].loc[0:2]
df1
```

```python
df2 = df[['id', 'age', 'gender']].loc[3:5]
df2
```

```python
df3 = pd.concat([df1, df2], axis=0) # Concat by rows
df3
```

```python
df4 = pd.concat([df2, df1], axis=0) # Concat by rows
df4
```

# Merge Examples

```python
patient_df = df[['id', 'age', 'gender']].loc[0:5]
patient_df
```

```python
lab_df = df[['id', 'chol', 'weight', 'height']].loc[2:7]
lab_df
```

```python
temp = pd.concat([patient_df, lab_df])
```

```python
temp
```

## Inner Join

```python
result_df = pd.merge(patient_df, lab_df, how='inner', left_on='id', right_on='id')
result_df
```

## Left Join

```python
result_df = pd.merge(patient_df, lab_df, how='left', left_on='id', right_on='id')
result_df
```

## Right Join (useless)

```python
result_df = pd.merge(patient_df, lab_df, how='right', left_on='id', right_on='id')
result_df
```

```python
# You can achieve the same thing by simply switching the order of tables
result_df = pd.merge(lab_df, patient_df, how='left', left_on='id', right_on='id')
result_df
```

## Full (or Outer) Join

```python
result_df = pd.merge(lab_df, patient_df, how='outer', left_on='id', right_on='id')
result_df
```
