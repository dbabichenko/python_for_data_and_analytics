```python
df['Job Title'].unique()
```

```python
def determine_position_level(position):
    position = position.lower()
    if "senior" in position:
        return "senior"
    elif "entry" in position:
        return "entry level"
    else:
        return "mid level"
```

```python
df['Position_Level'] = df['Job Title'].apply(lambda x: determine_position_level(x))
df.head()
```

```python
df['Position_Level'].value_counts().plot.bar()
```
