## K-Nearest Neighbor (KNN)



* This KNN tutorial contains code from "Evaluating a Classification Model" post available at http://www.ritchieng.com/machine-learning-evaluate-classification-model/

## The Data

This dataset is originally from the National Institute of Diabetes and Digestive and Kidney Diseases. The objective of the dataset is to diagnostically predict whether or not a patient has diabetes, based on certain diagnostic measurements included in the dataset. Several constraints were placed on the selection of these instances from a larger database. In particular, all patients here are females at least 21 years old of Pima Indian heritage.

https://www.kaggle.com/uciml/pima-indians-diabetes-database/version/1#


```python
# read the data into a Pandas DataFrame
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn

%matplotlib inline

df = pd.read_csv('pima_indians_diabetes.csv')
df.head()
```

```python
# define X and y
X = df[['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age']]

y = df['Outcome']
```

```python
# split X and y into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=1)
```

```python
from sklearn.neighbors import KNeighborsClassifier

# instantiate model
model = KNeighborsClassifier(n_neighbors=3)

# fit model
model.fit(X_train, y_train)
```

```python
# make class predictions for the testing set
y_pred_class = model.predict(X_test)
y_pred_class
```

**Classification accuracy**: percentage of correct predictions

```python
# calculate accuracy
from sklearn import metrics
print(metrics.accuracy_score(y_test, y_pred_class))
print(metrics.roc_auc_score(y_test, y_pred_class))
```

```python
from sklearn.metrics import RocCurveDisplay
RocCurveDisplay.from_estimator(model, X_test, y_test)
plt.plot([0, 1], [0, 1], 'k--')  # random-guess baseline
plt.show()
```

```python
k_values = [3, 5, 7, 9, 11, 13, 15, 17, 19]
for k in k_values:
    # instantiate model
    model = KNeighborsClassifier(n_neighbors=k)

    # fit model
    model.fit(X_train, y_train)
    y_pred_class = model.predict(X_test)
    print("Neighbors: ", k)
    print("Accuracy: ", metrics.accuracy_score(y_test, y_pred_class))
    print("AUC: ", metrics.roc_auc_score(y_test, y_pred_class))
```

```python
k_values = [3, 5, 7, 9, 11, 13, 15, 17, 19]
auc_values = []
accuracy_values = []

for k in k_values:
    # instantiate model
    model = KNeighborsClassifier(n_neighbors=k)

    # fit model
    model.fit(X_train, y_train)
    y_pred_class = model.predict(X_test)
    
    accuracy_values.append(metrics.accuracy_score(y_test, y_pred_class))
    auc_values.append(metrics.roc_auc_score(y_test, y_pred_class))
```

```python
print(k_values)
print(auc_values)
print(accuracy_values)
```

```python
result = pd.concat([pd.DataFrame(k_values), pd.DataFrame(auc_values), pd.DataFrame(accuracy_values)], axis=1)
result.columns = ['k', 'auc', 'accuracy']
result.head()
```

```python
sn.lineplot(x='k', y='auc', data=result)
```
