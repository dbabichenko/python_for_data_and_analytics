```python
from sklearn import preprocessing as prep
from sklearn.preprocessing import MinMaxScaler
```

```python
df['weight_scaled'] = prep.scale(df['weight'])
```

```python
sn.histplot(x=df['weight_scaled'], kde=True)
```

```python
scaler = MinMaxScaler()
df['weight_min_max'] = scaler.fit_transform(df[['weight']])
```

```python
sn.histplot(x=df['weight_min_max'], kde=True)
```
